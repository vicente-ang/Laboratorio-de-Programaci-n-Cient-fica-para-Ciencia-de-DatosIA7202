  <strong style="display:block; margin-bottom:8px;">📖 Unidad de observación</strong>

  La <strong>unidad de observación</strong> es el objeto que representa cada fila de una tabla. Puede ser una persona, una venta, una medición, una película o, en este caso, una versión de un personaje de cómic: Batman (LEGO) y Batman (Arkham) son filas distintas, con atributos propios.

  Identificarla correctamente es uno de los primeros pasos de cualquier análisis, porque determina <strong>qué estamos contando, comparando y transformando</strong>. Antes de modificar una tabla conviene preguntarse: ¿qué representa exactamente una fila?, ¿qué columna permite identificarla de forma única? y ¿cuántas observaciones esperamos conservar después de cada transformación?

  Por ejemplo, si nuestra unidad de observación es una versión de un personaje, cada versión debería aparecer una sola vez. Si dos filas describen a la misma versión, no estamos incorporando una nueva observación, sino duplicando información existente.

  Esto puede parecer un detalle menor, pero tiene efectos concretos. Dos filas idénticas tendrán distancia cero entre sí y, además, ese personaje quedará <strong>sobrerrepresentado</strong>: contará dos veces al calcular promedios, distribuciones o agrupamientos. En otras palabras, una duplicación puede alterar la estructura aparente de los datos.

  Por eso, antes de construir features o calcular similitudes, conviene verificar que la tabla respete la unidad de observación que queremos estudiar.

### 1.2 — Unidades y datos faltantes (0,4 puntos)

Ahora comenzaremos a transformar algunas columnas del catálogo en features numéricas utilizables.

Para eso debemos resolver dos problemas concretos: <strong>normalizar unidades</strong> y <strong>representar correctamente los datos faltantes</strong>. Alturas y pesos aparecen mezclando números, texto y distintas unidades, mientras que la ausencia de información puede estar codificada como <code>"-"</code>, <code>0</code> u otros valores especiales.

La idea será llevar cada magnitud a una unidad común y convertir los valores que realmente significan «sin dato» en <code>null</code>. Así, un número representará una medición conocida y no una ausencia disfrazada.

<!-- DEFINICIÓN -->
<div style="padding:16px 20px; margin:16px 0; border-left:4px solid #3fb950; background:rgba(63,185,80,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">📖 Extraer valores y normalizar unidades</strong>

  Cuando una columna mezcla números, texto y unidades, primero necesitamos <strong>separar la magnitud de su unidad</strong>. Para ello podemos usar <code>str.extract(patron, grupo)</code>, que recupera partes específicas de un texto mediante una <strong>expresión regular</strong>, un patrón que describe qué forma tiene el texto buscado. Cada par de paréntesis del patrón es un <strong>grupo de captura</strong>, numerado desde 1: en <code>r"([\d.]+) (kg|g)"</code>, el grupo 1 es el número (dígitos y puntos) y el grupo 2 la unidad. Si el patrón no calza, <code>extract</code> devuelve <code>null</code>.

  Una vez identificada la unidad, <code>pl.when(...).then(...).otherwise(...)</code> permite aplicar distintas conversiones según cada caso. Así podemos transformar representaciones diferentes de una misma magnitud a una <strong>unidad común</strong>.

  Veamos el mismo problema en una feria libre, donde los pesos vienen expresados en gramos o kilogramos.
</div>

````python
feria = pl.DataFrame(
    {
        "producto": ["palta", "tomate", "cilantro"],
        "peso": ["1.5 kg", "800 g", "-"],
    }
)
valor = pl.col("peso").str.extract(r"([\d.]+) (kg|g)", 1)
unidad = pl.col("peso").str.extract(r"([\d.]+) (kg|g)", 2)
feria.with_columns(
    pl.when(unidad == "kg")
    .then(valor.cast(pl.Float64) * 1000)
    .otherwise(valor.cast(pl.Float64))
    .alias("peso_g")
)
````

### 1.3 — Poderes como multi-hot (0,2 puntos)

<!-- DEFINICIÓN -->
<div style="padding:16px 20px; margin:16px 0; border-left:4px solid #3fb950; background:rgba(63,185,80,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">📖 Multi-hot y vocabulario aprendido</strong>

<strong>One-hot</strong> pone un 1 en la columna de la categoría de cada observación y 0 en las demás; <strong>multi-hot</strong> puede poner varios 1, uno por etiqueta. <code>MultiLabelBinarizer</code> aprende las etiquetas con <code>fit</code>, las guarda ordenadas en <code>classes_</code> y aplica esas mismas columnas con <code>transform</code>. <code>fit_transform</code> hace ambos pasos sobre los datos de ajuste. Ese conjunto de etiquetas aprendidas es el <strong>vocabulario</strong> del encoder. Una lista vacía deja todas las columnas en 0: expresa ausencia de etiquetas registradas.

Al llegar datos nuevos, el número y el significado de las columnas deben mantenerse. Por eso el encoder no se reajusta. Una etiqueta que no está en el vocabulario no tiene columna donde escribirse: <code>MultiLabelBinarizer</code> la descarta y emite una advertencia (<code>UserWarning</code>) que la nombra. La última línea de la celda siguiente lo muestra con una receta que trae merkén.
</div>

````python
from sklearn.preprocessing import MultiLabelBinarizer

recetas = [["palta", "tomate"], ["queso"], []]
encoder_recetas = MultiLabelBinarizer().fit(recetas)
print("Columnas:", encoder_recetas.classes_)
print(encoder_recetas.transform([["tomate"], []]))
print(encoder_recetas.transform([["tomate", "merkén"]]))
````

Construyan `poderes_explorados` a partir de `catalogo`, con `name`, `poderes` como lista de strings y `n_poderes` como la cantidad de poderes distintos. Eliminen etiquetas repetidas con `list.unique(maintain_order=True)` antes de contar: una etiqueta repetida no es otro poder. `superpowers` viene como `"['Flight', 'Super Strength']"`. El patrón `str.extract_all(r"'[^']+'")` toma cada texto entre comillas simples; después hay que quitarle las comillas con `list.eval(...)` (hint: `str.strip_chars`). No transformen filas con ciclos.

- [ ] Ajustar `binarizador` solo con las listas del catálogo y guardar su matriz en `poderes_multihot`.
- [ ] Definir `BINARIAS`, la lista de nombres de columna `poder_<etiqueta>`, en el orden de `binarizador.classes_` y conservando la escritura original de cada etiqueta.
- [ ] Construir `poderes_codificados_explorados`: conservar `name`, `poderes` y `n_poderes` y agregar el bloque multi-hot como `Int8` con `DataFrame.hstack` de Polars.
- [ ] Guardar `poderes_binarios`, la misma matriz como booleana, para comparar distancias después.

Cada fila debe seguir correspondiendo a la misma versión del personaje. La matriz debe tener 1 367 filas y 295 columnas; cada valor es 0 o 1. La lista vacía genera una fila de ceros. Los nombres de columnas vienen del encoder, no de una lista definida a mano.

````python
poderes_explorados = catalogo.select(
    "name",
    (
        pl.col("superpowers")
        .str.extract_all(r"'[^']+'")
        .list.eval(pl.element().str.strip_chars("'"))
        .list.unique(maintain_order=True)
        .alias("poderes")
    ),
).with_columns(
    pl.col("poderes").list.len().cast(pl.Int64).alias("n_poderes")
)

binarizador = MultiLabelBinarizer()
poderes_multihot = binarizador.fit_transform(
    poderes_explorados["poderes"].to_list()
)
BINARIAS = [f"poder_{etiqueta}" for etiqueta in binarizador.classes_]

bloque_poderes = pl.DataFrame(
    poderes_multihot,
    schema=BINARIAS,
    orient="row",
).cast(pl.Int8)

poderes_codificados_explorados = poderes_explorados.hstack(bloque_poderes)

poderes_binarios = poderes_multihot.astype(bool)
````

````python
print(f"Multi-hot: {poderes_multihot.shape}")
print(f"Fracción de unos: {poderes_multihot.mean():.3f}")
assert poderes_explorados["name"].equals(catalogo["name"])
assert poderes_multihot.shape == (1367, 295)
assert bloque_poderes.null_count().sum_horizontal().item() == 0
assert set(bloque_poderes.to_numpy().ravel()) <= {0, 1}
````

  <strong style="display:block; margin-bottom:8px;">📖 Transformaciones de escala</strong>
  Sea $x$ una columna, con media $\bar{x}$, desviación estándar $s$, mínimo $x_{\min}$, máximo $x_{\max}$, mediana $\tilde{x}$ y rango intercuartílico $\mathrm{IQR}$ (percentil 75 menos percentil 25).
  <ul>
  <li><strong>Estandarización</strong> (<code>StandardScaler</code>): $z = (x - \bar{x}) / s$. Deja media 0 y desviación 1, pero $\bar{x}$ y $s$ se mueven mucho con un solo valor extremo.</li>
  <li><strong>Min-max</strong> (<code>MinMaxScaler</code>): $(x - x_{\min}) / (x_{\max} - x_{\min})$. Deja todo entre 0 y 1; si el máximo es extremo, el resto queda apretado cerca de 0.</li>
  <li><strong>Escalamiento robusto</strong> (<code>RobustScaler</code>): $(x - \tilde{x}) / \mathrm{IQR}$. Usa mediana e IQR, menos sensibles a valores extremos que media y desviación estándar.</li>
  <li><strong>Transformación logarítmica</strong>: $\log(1 + x)$. Comprime las colas largas: 10 y 10 000 pasan a 2,4 y 9,2. Se suma 1 para que un 0 no dé $-\infty$; en NumPy es <code>np.log1p</code>.</li>
  </ul>
  Ninguna es la correcta en general: la elección depende de la distribución de cada feature.
</div>

<!-- PREGUNTA -->
<div style="padding:16px 20px; margin:16px 0; border-left:4px solid #58a6ff; background:rgba(88,166,255,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">❓ Elección antes de transformar</strong>
  1. A partir de las distribuciones, ¿qué transformación proponen para los puntajes y cuál para las medidas?
2. ¿Por qué min-max puede comprimir casi todos los pesos cerca de 0? ¿Qué problema de los valores extremos no resuelve un escalamiento robusto?
</div>

**1.** Usaría mediana y `StandardScaler` para puntajes acotados de 0 a 100; para las medidas, mediana, `log1p` y `RobustScaler`, porque presentan colas largas.

**2.** Un peso cercano a 99 millones de kg haría que min-max concentrara casi todos los otros cerca de cero. `RobustScaler` resiste ese extremo, pero no lo comprime: eso lo hace `log1p`.

<!-- DEFINICIÓN -->
<div style="padding:16px 20px; margin:16px 0; border-left:4px solid #3fb950; background:rgba(63,185,80,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">📖 Pipeline: una secuencia de transformaciones</strong>
  Un <strong><code>Pipeline</code></strong> encadena operaciones: al ajustar, cada transformador aprende con la salida del anterior; al transformar datos nuevos, aplica los parámetros ya aprendidos en el mismo orden. <code>SimpleImputer</code> aprende una mediana, <code>StandardScaler</code> aprende media y desviación, y <code>RobustScaler</code> mediana e IQR. <code>log1p</code> no aprende parámetros; <code>FunctionTransformer</code> permite incorporarlo a la secuencia, y <code>feature_names_out="one-to-one"</code> le indica que cada columna conserva su nombre. <code>set_output(transform="polars")</code> hace que la salida sea un DataFrame de Polars con nombres, en vez de un arreglo de NumPy.
</div>

````python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import FunctionTransformer, StandardScaler

consumos = pl.DataFrame({"consumo_kwh": [80.0, None, 150.0, 900.0]})
pipeline_consumo = Pipeline(
    [
        ("imputar", SimpleImputer(strategy="median")),
        ("log", FunctionTransformer(np.log1p, feature_names_out="one-to-one")),
        ("escalar", StandardScaler()),
    ]
).set_output(transform="polars")
print(pipeline_consumo.fit_transform(consumos))
print(pipeline_consumo.transform(pl.DataFrame({"consumo_kwh": [120.0]})))
````

Construyan `pipeline_puntajes` con mediana y `StandardScaler`, y `pipeline_medidas` con mediana, `log1p` y `RobustScaler`. Ajusten ambos solo con el catálogo. Guarden sus salidas en `puntajes_transformados` y `medidas_transformadas`, como DataFrames de Polars con los nombres originales.

Construyan además `X_estandar` aplicando `StandardScaler` a `X_numerico`, y `X_log_robusto` pasando las dos salidas anteriores a arreglos (`to_numpy()`) y uniéndolas con `np.hstack`, en el orden de `NUMERICAS`. Cada matriz debe tener 1 367 filas y nueve columnas, sin faltantes. Las compararemos en la etapa 2.

En la 5.4 combinaremos los puntajes con el texto de los poderes. Para eso, construyan también `puntajes_unitarios_catalogo`.

- [ ] Dividan cada fila de `puntajes_transformados` por su norma euclidiana $\lVert x \rVert_2 = \sqrt{\sum_j x_j^2}$, donde $x_j$ es el puntaje $j$ de esa fila. Por ejemplo, $(3, 4)$ tiene norma 5 y pasa a $(0{,}6;\ 0{,}8)$.
- [ ] Si la norma de una fila es 0, la fila debe quedar en ceros: calculen las normas con `np.linalg.norm(..., axis=1, keepdims=True)` y usen `.clip(1e-9)` en el denominador para no dividir por 0.

Ojo con la diferencia: los escaladores anteriores transforman **columnas**; esta operación normaliza **filas**. Conserva la dirección del vector y descarta su largo, así que dos perfiles proporcionales quedan iguales. Como los puntajes ya están centrados y escalados, la dirección describe en qué se destaca un personaje respecto del resto del catálogo, no su nivel absoluto de capacidad.

````python
from sklearn.preprocessing import RobustScaler

pipeline_puntajes = Pipeline(
    [
        ("imputar", SimpleImputer(strategy="median")),
        ("escalar", StandardScaler()),
    ]
).set_output(transform="polars")
puntajes_transformados = pipeline_puntajes.fit_transform(features.select(PUNTAJES))

pipeline_medidas = Pipeline(
    [
        ("imputar", SimpleImputer(strategy="median")),
        (
            "log",
            FunctionTransformer(np.log1p, feature_names_out="one-to-one"),
        ),
        ("escalar", RobustScaler()),
    ]
).set_output(transform="polars")
medidas_transformadas = pipeline_medidas.fit_transform(features.select(MEDIDAS))

X_estandar = StandardScaler().fit_transform(X_numerico)

X_log_robusto = np.hstack(
    [puntajes_transformados.to_numpy(), medidas_transformadas.to_numpy()]
)

puntajes_array = puntajes_transformados.to_numpy()
normas = np.linalg.norm(puntajes_array, axis=1, keepdims=True)
puntajes_unitarios_catalogo = puntajes_array / normas.clip(1e-9)
````

## Etapa 2 — Distancias y vecinos (0,8 puntos)

<!-- LORE -->
<div style="padding:16px 20px; margin:16px 0; border-left:4px solid #0f9e99; background:rgba(15,158,153,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">📜 Groot y los robots</strong>
  Con las features ya numéricas, la Magneta probó su primera búsqueda: «personajes parecidos a Groot». Le salieron Bumblebee, Iron Monger y Soundwave. «Son todos robots de varias toneladas», escribió, «Groot es un árbol». Tiene razón en sospechar: los números están bien, pero no todos pesan lo mismo en la comparación.
</div>

<!-- DEFINICIÓN -->
<div style="padding:16px 20px; margin:16px 0; border-left:4px solid #3fb950; background:rgba(63,185,80,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">📖 Distancia euclidiana y vecinos más cercanos</strong>
  Si cada personaje es un vector $x = (x_1, \dots, x_p)$ con $p$ features, la <strong>distancia euclidiana</strong> entre $x$ e $y$ es

  $$d(x, y) = \sqrt{\sum_{j=1}^{p} (x_j - y_j)^2}.$$

  Los <strong>vecinos más cercanos</strong> (<em>nearest neighbors</em>) de un personaje son los que tienen menor distancia a él. Como cada término es una diferencia al cuadrado, una feature cuyo rango es de miles domina la suma frente a otra cuyo rango es de decenas. Pensemos en departamentos descritos por precio en pesos y número de dormitorios: una diferencia de un dormitorio aporta 1 a la suma, y una de un millón de pesos aporta $10^{12}$.
</div>

````python
import numpy as np
from sklearn.neighbors import NearestNeighbors

departamentos = np.array(
    [
        [95_000_000, 2],  # consulta
        [96_000_000, 5],  # precio casi igual, tres dormitorios más
        [110_000_000, 2],  # mismos dormitorios, más caro
    ]
)
_, indices = (
    NearestNeighbors(n_neighbors=2)
    .fit(departamentos)
    .kneighbors(departamentos[:1])
)
print("Vecino del primer departamento:", indices[0][1])
````

### 2.2 — Vecinos sin escalar (0,2 puntos)

Completen `vecinos_de`: debe devolver un DataFrame con `name` y `distancia` de los `k` vecinos más cercanos de `consulta` en la matriz `X`, ordenados por distancia y sin incluir a la consulta. Las filas de `X` siguen el orden de `nombres`, la lista que construimos en la 1.6 (hint: `NearestNeighbors` y su método `kneighbors`; pidan `k + 1` vecinos, porque el más cercano es la propia consulta).

````python
def vecinos_de(X, consulta: str, k: int = 5) -> pl.DataFrame:
    """Busca vecinos euclidianos en la matriz X sin incluir la consulta."""
    if consulta not in nombres:
        raise KeyError(f"{consulta!r} no está en nombres")

    indice_consulta = nombres.index(consulta)
    modelo = NearestNeighbors(
        n_neighbors=min(k + 1, len(nombres)),
        metric="euclidean",
    ).fit(X)

    distancias, indices = modelo.kneighbors(
        X[indice_consulta : indice_consulta + 1]
    )
    indices = indices[0]
    distancias = distancias[0]

    mascara = indices != indice_consulta
    indices = indices[mascara][:k]
    distancias = distancias[mascara][:k]

    nombres_array = np.asarray(nombres, dtype=object)
    return pl.DataFrame(
        {
            "name": nombres_array[indices].tolist(),
            "distancia": distancias.astype(float).tolist(),
        }
    )
````

Los cinco vecinos de Groot sobre `X_numerico`, junto con los datos que permiten juzgarlos:

````python
vecinos_crudos = vecinos_de(X_numerico, "Groot")
vecinos_crudos.join(features, on="name", maintain_order="left").select(
    "name", "distancia", "strength_score", "altura_cm", "peso_kg"
)
````

### 2.5 — Poderes: euclidiana o Jaccard (0,2 puntos)

<!-- DEFINICIÓN -->
<div style="padding:16px 20px; margin:16px 0; border-left:4px solid #3fb950; background:rgba(63,185,80,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">📖 Distancia de Jaccard</strong>
  Para dos conjuntos $A$ y $B$, la <strong>distancia de Jaccard</strong> es

  $$d_J(A, B) = 1 - \frac{|A \cap B|}{|A \cup B|}.$$

  Con vectores binarios, la distancia euclidiana al cuadrado cuenta las posiciones en que difieren. Ninguna de las dos distancias cambia si agregamos columnas donde ambos valen 0; la diferencia es que Jaccard divide por la cantidad de elementos presentes en alguno de los dos. Por ejemplo, $A = \{\text{vuelo}\}$ y $B = \{\text{vuelo}, \text{fuerza}\}$ difieren en un poder: euclidiana al cuadrado 1 y Jaccard $1 - 1/2 = 0{,}5$. Dos personajes que comparten 19 poderes y difieren en uno también tienen euclidiana al cuadrado 1, pero Jaccard $1 - 19/20 = 0{,}05$. La euclidiana cuenta diferencias; Jaccard las pone en proporción. Un churrasco italiano (pan, carne, palta, tomate, mayo) y un barros luco (pan, carne, queso) comparten 2 de 6 ingredientes: $d_J = 1 - 2/6 \approx 0{,}67$.
</div>

Para dos conjuntos vacíos, la fórmula tiene denominador 0. scikit-learn sigue la convención de SciPy: distancia Jaccard 0 entre dos vectores vacíos. Es una decisión computacional, no evidencia de que sus capacidades reales sean iguales.

Usen `poderes_binarios`, construido en la 1.3. Con `vecinos_mas_cercanos`, muestren los cinco vecinos de Deadpool con la métrica euclidiana y con `metrica="jaccard"`, y cuenten cuántos personajes tienen las 295 columnas en 0.

````python
for metrica in ["euclidean", "jaccard"]:
    print(metrica)
    display(vecinos_mas_cercanos(
        poderes_binarios.astype(bool), nombres, "Deadpool", metrica=metrica
    ))
print("Sin poderes registrados:", int((poderes_binarios.sum(axis=1) == 0).sum()))
````

<!-- DEFINICIÓN -->
<div style="padding:16px 20px; margin:16px 0; border-left:4px solid #3fb950; background:rgba(63,185,80,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">📖 Tokens y bag of words</strong>
  <strong>Tokenizar</strong> es partir un texto en unidades, los <strong>tokens</strong>: palabras, trozos de palabras o signos. El <strong>bag of words</strong> representa cada texto con un vector que cuenta cuántas veces aparece cada palabra del vocabulario, sin importar el orden. <code>CountVectorizer</code> lo construye; las <strong>stop words</strong> son palabras muy frecuentes, como «el» o «de», que ayudan poco a distinguir un texto de otro y conviene descartar. scikit-learn solo trae la lista del inglés (<code>stop_words="english"</code>), así que en el ejemplo la pasamos a mano. Como casi todas las celdas del resultado valen 0, <code>CountVectorizer</code> lo devuelve como <strong>sparse matrix</strong> (matriz dispersa), que solo guarda las celdas distintas de 0; <code>.toarray()</code> la convierte en un arreglo común para mirarla.
</div>

````python
from sklearn.feature_extraction.text import CountVectorizer

resenas = [
    "el completo italiano estaba rico",
    "el completo estaba frío",
    "la micro venía llena",
]
contador = CountVectorizer(stop_words=["el", "la", "estaba", "venía"])
conteos = contador.fit_transform(resenas)
pl.DataFrame(conteos.toarray(), schema=list(contador.get_feature_names_out()))
````

<!-- DEFINICIÓN -->
<div style="padding:16px 20px; margin:16px 0; border-left:4px solid #3fb950; background:rgba(63,185,80,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">📖 TF-IDF</strong>
  <strong>TF-IDF</strong> pondera cada conteo por lo informativa que es la palabra. Para la palabra $t$ en el texto $d$, con $N$ textos en total y $\mathrm{df}(t)$ textos que contienen $t$,

  $$\mathrm{tfidf}(t, d) = \mathrm{tf}(t, d) \cdot \left(\ln \frac{1 + N}{1 + \mathrm{df}(t)} + 1\right),$$

  donde $\mathrm{tf}(t, d)$ es el conteo. Una palabra que está en todos los textos recibe el factor mínimo, 1: conserva su conteo, pero no gana peso. <code>TfidfVectorizer</code> normaliza cada fila que conserva al menos un token para que su norma euclidiana sea 1; las que pierden todos sus tokens quedan en ceros. Igual que los conteos, el resultado es una sparse matrix.
</div>

<!-- DEFINICIÓN -->
<div style="padding:16px 20px; margin:16px 0; border-left:4px solid #3fb950; background:rgba(63,185,80,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">📖 Similitud coseno</strong>
  La <strong>similitud coseno</strong> entre $a$ y $b$ es $\cos(a, b) = \frac{a \cdot b}{\lVert a \rVert\,\lVert b \rVert}$, y la distancia coseno es $1 - \cos(a, b)$. Mide el ángulo entre dos vectores, no su largo: una descripción larga y otra corta que usan las palabras en las mismas proporciones quedan a distancia 0. Si los vectores tienen norma 1, como las filas no vacías de TF-IDF, la similitud es simplemente el producto punto $a \cdot b$.
</div>

````python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

pesador = TfidfVectorizer(stop_words=["el", "la", "estaba", "venía"])
pesos = pesador.fit_transform(resenas)
print("Coseno con conteos:")
print(np.round(cosine_similarity(conteos), 2))
print("Coseno con TF-IDF:")
print(np.round(cosine_similarity(pesos), 2))
````

Las dos primeras reseñas solo comparten `completo`. Con conteos su similitud es 0,41; con TF-IDF baja a 0,29, porque `completo` aparece en dos de las tres reseñas y pesa menos que `italiano`, `rico` o `frío`, que están en una sola.

### 4.1 — PCA de los puntajes con loadings (0,4 puntos)

<!-- DEFINICIÓN -->
<div style="padding:16px 20px; margin:16px 0; border-left:4px solid #3fb950; background:rgba(63,185,80,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">📖 PCA y loadings</strong>
  El <strong>análisis de componentes principales</strong> (PCA) construye nuevos ejes, las <strong>componentes</strong>, que son combinaciones lineales de las features originales. La primera apunta hacia donde los datos varían más; cada una de las siguientes es perpendicular (ortogonal) a las anteriores y recoge la mayor variación que queda. <code>components_</code> contiene los coeficientes de cada combinación. La <strong>varianza explicada</strong> indica qué fracción de la varianza total, la suma de las varianzas de todas las features, recoge cada componente. Al conservar solo dos componentes, las distancias y vecindades pueden cambiar.

  Los <strong>loadings</strong> muestran cómo se relaciona cada variable original con las componentes. Se calculan multiplicando los coeficientes de cada componente por la raíz de su varianza. Con features estandarizadas, como nuestros puntajes, cada loading es aproximadamente la correlación entre la variable y la componente, así que queda entre −1 y 1. En un mapa se dibujan como flechas desde el origen: el signo indica la dirección de asociación y el valor absoluto, su intensidad en cada eje. Una flecha corta en este plano puede corresponder a una variable mejor representada por otras componentes. Los loadings ayudan a interpretar los ejes; no prueban que existan grupos.
</div>

Ajusten `pca_puntajes`, un `PCA` sobre las seis columnas de puntajes de `tabular`, ya imputadas y estandarizadas. La celda que sigue a la suya muestra la varianza explicada de las seis componentes.

````python
from sklearn.decomposition import PCA

pca_puntajes = PCA().fit(tabular.select(PUNTAJES).to_numpy())
````

````python
print(
    "Varianza explicada:", np.round(pca_puntajes.explained_variance_ratio_, 3)
)
````

  <strong style="display:block; margin-bottom:8px;">📖 Representación y agrupamiento</strong>

  La pregunta «¿aparecen grupos naturales?» tiene nombre: <strong>clustering</strong>. Un algoritmo de clustering recibe los vectores de features, sin ninguna etiqueta, y los reparte en grupos de modo que los puntos de un mismo grupo queden cerca entre sí. Su resultado depende de dos decisiones.

  Depende, primero, de la <strong>representación</strong>. El algoritmo agrupa puntos, no personajes: si cambiamos las features, cambian las posiciones, y con ellas los grupos. Con solo fuerza y velocidad, Superman y Hulk podrían caer juntos; si agregamos inteligencia, quizás ya no.

  Depende, además, del <strong>algoritmo</strong>, porque cada uno tiene su propia idea de qué es un grupo. En la figura, cada fila es un mismo conjunto de puntos y cada columna, un algoritmo distinto; el color indica el grupo asignado y el negro, los puntos que el algoritmo deja sin grupo. En la primera fila, K-Means corta los dos anillos por la mitad, porque busca grupos compactos alrededor de un centro, mientras que DBSCAN sigue las zonas densas y separa cada anillo. En la última, los puntos están repartidos al azar y no hay grupos, pero K-Means, Ward o Gaussian Mixture igual entregan tres.

  <div style="text-align: center; margin: 16px 0;">
    <img src="static/sphx_glr_plot_cluster_comparison_001.png" alt="Comparación de once algoritmos de clustering de scikit-learn sobre seis conjuntos de puntos en dos dimensiones: dos anillos concéntricos, dos medias lunas, tres grupos de distinta dispersión, tres grupos alargados, tres grupos redondos y puntos uniformes sin estructura. Cada algoritmo colorea los grupos que encuentra y los resultados difieren entre columnas." style="width: 900px; max-width: 100%; height: auto;">
    <br>
    <small>Fuente: documentación de scikit-learn, «Comparing different clustering algorithms on toy datasets».</small>
  </div>

  Que un algoritmo entregue grupos no significa que existan. Por eso, en esta etapa tendremos que justificar cada agrupación con evidencia.

  <br>
</div>

<!-- DEFINICIÓN -->
<div style="padding:16px 20px; margin:16px 0; border-left:4px solid #3fb950; background:rgba(63,185,80,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">📖 K-Means, inercia y método del codo</strong>
  <strong>K-Means</strong> reparte los puntos en $k$ grupos y busca los centros $\mu_1, \dots, \mu_k$ que minimizan la <strong>inercia</strong>, la suma de distancias al cuadrado de cada punto a su centro: $\sum_i \lVert x_i - \mu_{c(i)} \rVert^2$, donde $c(i)$ es el grupo asignado al punto $x_i$. Para lograrlo alterna dos pasos hasta que las asignaciones dejan de cambiar: asigna cada punto al centro más cercano y mueve cada centro al promedio de sus puntos.

  La inercia siempre baja al aumentar $k$; el <strong>método del codo</strong> busca el $k$ desde el cual deja de bajar mucho. K-Means parte de centros al azar, así que su resultado depende de la semilla. <code>n_init</code> indica cuántas veces parte desde centros distintos: con <code>n_init=10</code> se queda con la corrida de menor inercia, y con <code>n_init=1</code> queda expuesto a lo que salga de una sola semilla.
</div>

<!-- DEFINICIÓN -->
<div style="padding:16px 20px; margin:16px 0; border-left:4px solid #3fb950; background:rgba(63,185,80,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">📖 Silhouette y ARI</strong>
  Para un punto, sea $a$ su distancia media a los demás puntos de su grupo y $b$ la menor distancia media a los puntos de otro grupo. Su <strong>silhouette</strong> es $s = (b - a) / \max(a, b)$, entre −1 y 1; el coeficiente es el promedio sobre todos los puntos. Cerca de 1, los grupos están bien separados; cerca de 0, el punto está en la frontera entre dos grupos; negativo, quedaría mejor en otro grupo.

  El <strong>índice de Rand ajustado</strong> (ARI) compara dos particiones de los mismos puntos: vale 1 si coinciden aunque numeren los grupos distinto, y cerca de 0 si coinciden como lo harían al azar; puede ser negativo si coinciden menos que eso. Sirve para medir estabilidad (la misma representación con otra semilla) y acuerdo (dos representaciones distintas), sin usar ninguna etiqueta verdadera.
</div>

````python
from sklearn.metrics import adjusted_rand_score

print(adjusted_rand_score([0, 0, 1, 1, 2, 2], [2, 2, 0, 0, 1, 1]))
print(adjusted_rand_score([0, 0, 1, 1, 2, 2], [0, 1, 2, 0, 1, 2]))
````

### 5.2 — Arquetipos a partir de los puntajes (0,3 puntos)

Guarden en `X_puntajes` las seis columnas de puntajes de `tabular` como arreglo. Para cada `k` entre 2 y 6, construyan una fila de `tabla_k_puntajes` con las columnas `k`, `silhouette` y `estabilidad`:

- [ ] `silhouette`: el de `KMeans(n_clusters=k, n_init=10, random_state=0)` sobre `X_puntajes`, igual que en la 5.1;
- [ ] `estabilidad`: ajusten K-Means con `n_init=1` y las semillas 0 a 4, y promedien el ARI entre los 10 pares de particiones (hint: `itertools.combinations`).

````python
from itertools import combinations
from sklearn.metrics import adjusted_rand_score

X_puntajes = tabular.select(PUNTAJES).to_numpy()
filas_puntajes = []
for k in range(2, 7):
    modelo = KMeans(n_clusters=k, n_init=10, random_state=0)
    etiquetas = modelo.fit_predict(X_puntajes)
    semillas = [
        KMeans(n_clusters=k, n_init=1, random_state=s).fit_predict(X_puntajes)
        for s in range(5)
    ]
    acuerdo = np.mean([
        adjusted_rand_score(a, b) for a, b in combinations(semillas, 2)
    ])
    filas_puntajes.append({
        "k": k,
        "silhouette": float(silhouette_score(X_puntajes, etiquetas)),
        "estabilidad": float(acuerdo),
    })
tabla_k_puntajes = pl.DataFrame(filas_puntajes)
display(tabla_k_puntajes)
````

Para el entregable, los arquetipos de los personajes nuevos se predicen con el K-Means ya ajustado, sin reajustarlo.

Los equivalentes se buscan en una representación que junta dos bloques por personaje $i$: $t_i$, su fila TF-IDF, que tiene norma 1 si conserva tokens, y $s_i$, sus puntajes normalizados en la 1.6 (`puntajes_unitarios_catalogo`), multiplicados por 0,5. El texto aporta los poderes descritos y los puntajes, en qué se destaca el personaje.

¿Por qué 0,5? Es una elección de diseño. Como el coseno multiplica los vectores entre sí, el peso aparece al cuadrado: para dos personajes con ambos bloques de norma 1, el producto punto combinado es $t_i \cdot t_j + 0{,}25\, s_i \cdot s_j$, y la similitud coseno divide esa suma por $1 + 0{,}25 = 1{,}25$. El texto pesa entonces cuatro veces más que los puntajes en esa suma, aunque cuánto aporta cada bloque depende de cada par.

Dos límites: si el texto de un personaje es un vector cero, solo aporta su bloque de puntajes; y 0,5 no es un «50 % de importancia» ni garantiza equivalentes válidos. Como `tfidf` es una sparse matrix, los dos bloques se unen con `hstack` de `scipy.sparse`, no con el de NumPy.

````python
from scipy.sparse import csr_matrix, hstack

nuevos_arquetipos = arquetipos.predict(
    nuevos_tabular.select(PUNTAJES).to_numpy()
)
print("Nuevos por arquetipo:", np.bincount(nuevos_arquetipos))

puntajes_unitarios = puntajes_unitarios_catalogo[
    features["name"].is_in(nombres_texto).to_numpy()
]
representacion_juego = hstack(
    [tfidf, csr_matrix(0.5 * puntajes_unitarios)]
).tocsr()
base = con_texto.select("name", "creator")
for favorito in ["Deadpool", "Wolverine", "Hulk", "Iron Man", "Magneto"]:
    print(
        favorito, equivalentes(representacion_juego, base, favorito, k=3).rows()
    )
````

**DBSCAN** conecta regiones densas. Un punto es núcleo si tiene al menos `min_samples` puntos a distancia menor o igual que `eps`, incluyéndose a sí mismo. Un punto frontera está a distancia menor o igual que `eps` de un núcleo, pero no alcanza ese mínimo; los restantes son ruido, con etiqueta −1. No se fija `k`. El ejemplo provisto permite inspeccionar puntajes atípicos; sus parámetros no son universales y una etiqueta de ruido no prueba un error en la fuente.

````python
from sklearn.cluster import DBSCAN

ruido = DBSCAN(eps=1.0, min_samples=5).fit(X_puntajes).labels_ == -1
print(f"Personajes como ruido: {int(ruido.sum())}")
features.filter(pl.Series(ruido)).select("name", *PUNTAJES).head(10)
````
