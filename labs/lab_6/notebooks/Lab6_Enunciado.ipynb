### 1.3 — Auditoría y claves (0,4 puntos)

<!-- DEFINICIÓN -->
<div style="padding:16px 20px; margin:16px 0; border-left:4px solid #3fb950; background:rgba(63,185,80,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">📖 Análisis Exploratorio de Datos (EDA)</strong>

  El <strong>análisis exploratorio de datos</strong> (<em>Exploratory Data Analysis</em>, EDA) reúne evidencia antes de construir transformaciones, indicadores o variables para un modelo. Permite saber qué representa cada fuente, cómo está estructurada y qué problemas pueden cambiar la interpretación de sus resultados. No corrige todavía los datos: fundamenta las reglas que después aplicarán Silver y Gold.

  Un EDA debería revisar, al menos:

  <ul style="margin:10px 0 0 20px;">
    <li><strong>Dimensiones:</strong> número de filas y columnas.</li>
    <li><strong>Esquema:</strong> nombres de columnas y tipos de datos.</li>
    <li><strong>Unidad de observación:</strong> qué representa una fila.</li>
    <li><strong>Claves:</strong> columnas que identifican registros y posibles duplicados.</li>
    <li><strong>Valores ausentes:</strong> cantidad y distribución de nulos.</li>
    <li><strong>Distribuciones:</strong> rangos, valores frecuentes y posibles valores atípicos.</li>
    <li><strong>Categorías:</strong> valores posibles en variables discretas, como estados o tipos.</li>
    <li><strong>Relaciones entre fuentes:</strong> claves compartidas, cardinalidades y posibles registros huérfanos.</li>
  </ul>

  <p style="margin-top:10px;">
    El resultado del EDA debe permitir proponer contratos, anticipar qué uniones pueden multiplicar filas y definir qué registros entrarán a las variables RFM.
  </p>

</div>

A continuación, identifiquen qué contiene cada fuente de datos. Usen la función
`generar_eda` para hacerlo.

````python
from IPython.display import display
import polars as pl


def generar_eda(
    tabla: pl.DataFrame,
    nombre_fuente: str,
    n_muestra: int = 8,
    seed: int = 7202,
) -> None:
    """Muestra una exploración inicial de la fuente."""

    print(f"EDA rápido {nombre_fuente}\n")
    print("-" * 80)
    print(f"Muestra aleatoria - Semilla = {seed}")
    print("-" * 80)
    display(
        tabla.sample(
            n=min(n_muestra, tabla.height),
            seed=seed,
            shuffle=True,
        )
    )

    print("\n" + "-" * 80)
    print("Perfilamiento por columnas")
    print("-" * 80)

    nulos = tabla.null_count().transpose(
        include_header=True,
        header_name="columna",
        column_names=["nulos"],
    )

    distintos = tabla.select(pl.all().n_unique()).transpose(
        include_header=True,
        header_name="columna",
        column_names=["valores_distintos"],
    )

    esquema = pl.DataFrame(
        {
            "columna": tabla.columns,
            "tipo": [str(tipo) for tipo in tabla.schema.values()],
        }
    )

    if tabla.is_empty():
        porcentaje_nulos = pl.lit(None, dtype=pl.Float64)
    else:
        porcentaje_nulos = (pl.col("nulos") / tabla.height * 100).round(2)

    display(
        esquema.join(nulos, on="columna")
        .join(distintos, on="columna")
        .with_columns(
            porcentaje_nulos.alias("porcentaje_nulos")
        )
        .sort("porcentaje_nulos", descending=True)
    )

    print("\n\n" + "-" * 80)
    print("Estadísticas descriptivas")
    print("-" * 80)
    display(tabla.describe())
````

  El <strong>grano</strong> declara qué representa una fila: una orden, un ítem de una orden, un pago o una persona. La <strong>cardinalidad</strong> describe cuántas filas de una tabla pueden corresponder a una fila de otra: uno a uno, uno a muchos o muchos a muchos. Antes de hacer un <em>join</em>, comparen ambos granos; una relación muchos a muchos puede multiplicar filas y duplicar montos.
</div>

````python
generar_eda(orders_raw, "orders", n_muestra=10, seed=7202)
````

**2. Qué cambia al mezclar granos y por qué `customer_id` no basta**

Unir `order_items` con `payments` directamente por `order_id` produce un producto cartesiano dentro de cada orden: una orden con 3 ítems y 2 pagos genera 6 filas. Medido sobre estos datos, ese join pasa de 112.650 a **117.601 filas**, y la suma de `payment_value` salta de **16.008.872,12 a 20.308.134,71**, un 27 % de inflación. Por eso los pagos deben agregarse a grano de orden *antes* de cualquier unión hacia ítems o hacia cliente.

`customer_id` no sirve para medir frecuencia de compra porque, como muestra la evidencia anterior, se emite uno nuevo por orden: al agrupar por `customer_id` toda persona tendría `frequency = 1` por construcción y Recency sería siempre la de su única compra. La frecuencia solo es observable agrupando por `customer_unique_id`; con esa identidad, 2.997 personas tienen más de una orden y el RFM deja de ser degenerado.

  <strong style="display:block; margin-bottom:8px;">📖 Joins en Polars</strong>
  Un <em>join</em> relaciona filas por el valor de una llave, no por su posición. En Polars, la tabla sobre la que se llama <code>.join()</code> es la tabla izquierda: <code>izquierda.join(derecha, on="order_id", how="...")</code>.

  Sean \(A\) y \(B\) los conjuntos de llaves de las tablas izquierda y derecha. La cardinalidad importa: unir una orden con sus ítems es uno a muchos; si ambas tablas tienen varias filas por la misma llave, el resultado puede multiplicar filas y montos.

  <ul style="margin:10px 0 0 20px;">
    <li><code>how="inner"</code>: conserva las llaves de \(A \cap B\), presentes en ambas tablas. Es el valor por defecto.</li>
    <li><code>how="left"</code>: conserva todas las llaves de \(A\). Si una llave pertenece a \(A \setminus B\), las columnas de la derecha quedan nulas.</li>
    <li><code>how="right"</code>: conserva todas las llaves de \(B\). Si una llave pertenece a \(B \setminus A\), las columnas de la izquierda quedan nulas.</li>
    <li><code>how="full"</code>: conserva todas las llaves de \(A \cup B\). Las columnas del lado sin correspondencia quedan nulas.</li>
    <li><code>how="semi"</code>: conserva las filas de la izquierda con llaves en \(A \cap B\), pero devuelve solo columnas de la izquierda.</li>
    <li><code>how="anti"</code>: conserva las filas de la izquierda con llaves en \(A \setminus B\), pero devuelve solo columnas de la izquierda. Aquí sirve para detectar pagos cuya orden no existe y órdenes entregadas sin pago.</li>
  </ul>

#### 2.2.1 — Fechas, estados y clientes

Construyan `orders_silver` desde `orders_raw` y `customers_silver` desde
`customers_raw`. No modifiquen las tablas Bronze. En `orders_silver`:

- conviertan a `Datetime` `order_purchase_timestamp`, `order_approved_at`,
  `order_delivered_carrier_date`, `order_delivered_customer_date` y
  `order_estimated_delivery_date`;
- conserven los nulos originales; si un valor no nulo no se puede convertir,
  levanten `ContractViolation`;
- acepten solo `approved`, `canceled`, `created`, `delivered`, `invoiced`,
  `processing`, `shipped` y `unavailable`;
- agreguen `delivery_timestamp_missing`, verdadero únicamente para una orden
  `delivered` cuya fecha de entrega al cliente sea nula;
- rechacen una orden entregada cuya fecha de entrega sea anterior a la compra.

`customers_silver` conserva `customer_id` y `customer_unique_id` sin agrupar:
una persona puede aparecer mediante varios `customer_id`, y esa consolidación
corresponde a RFM, no a Silver.

````python
from src.pudulake.contracts import ContractViolation

COLUMNAS_FECHA = (
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date",
)
FORMATO_FECHA = "%Y-%m-%d %H:%M:%S"
ESTADOS_VALIDOS = {
    "approved",
    "canceled",
    "created",
    "delivered",
    "invoiced",
    "processing",
    "shipped",
    "unavailable",
}

# 1) Tipamos con strict=False para poder distinguir después qué falló.
orders_tipadas = orders_raw.with_columns(
    pl.col(columna)
    .str.strptime(pl.Datetime, format=FORMATO_FECHA, strict=False)
    .alias(columna)
    for columna in COLUMNAS_FECHA
)

# 2) Ausencia real vs. falla de conversión: original no nulo -> parseado nulo.
auditoria_fechas = pl.DataFrame(
    {
        "columna": list(COLUMNAS_FECHA),
        "nulos_originales": [
            orders_raw[columna].null_count() for columna in COLUMNAS_FECHA
        ],
        "no_interpretables": [
            (
                orders_raw[columna].is_not_null()
                & orders_tipadas[columna].is_null()
            ).sum()
            for columna in COLUMNAS_FECHA
        ],
    }
)
display(auditoria_fechas)

if auditoria_fechas["no_interpretables"].sum() > 0:
    raise ContractViolation(
        "silver.orders recibió texto no interpretable como fecha."
    )

# 3) Estados admitidos por el contrato.
estados_desconocidos = (
    set(orders_tipadas["order_status"].unique().to_list()) - ESTADOS_VALIDOS
)
if estados_desconocidos:
    raise ContractViolation(f"Estados fuera del contrato: {estados_desconocidos}")

# 4) Marca explícita de entrega sin timestamp y secuencia temporal.
orders_silver = orders_tipadas.with_columns(
    (
        (pl.col("order_status") == "delivered")
        & pl.col("order_delivered_customer_date").is_null()
    ).alias("delivery_timestamp_missing")
)

fuera_de_secuencia = orders_silver.filter(
    (pl.col("order_status") == "delivered")
    & (
        pl.col("order_delivered_customer_date")
        < pl.col("order_purchase_timestamp")
    )
).height
if fuera_de_secuencia:
    raise ContractViolation(
        f"{fuera_de_secuencia} órdenes entregadas antes de su compra."
    )

# 5) customers: el puente customer_id -> customer_unique_id, sin agrupar.
customers_silver = customers_raw.select("customer_id", "customer_unique_id")
if customers_silver["customer_id"].n_unique() != customers_silver.height:
    raise ContractViolation("silver.customers no es una fila por customer_id.")

display(
    orders_silver.select(
        pl.len().alias("ordenes"),
        pl.col("delivery_timestamp_missing").sum().alias("entregas_sin_fecha"),
        pl.col("order_purchase_timestamp").min().alias("primera_compra"),
        pl.col("order_purchase_timestamp").max().alias("ultima_compra"),
    )
)
display(
    customers_silver.select(
        pl.len().alias("filas"),
        pl.col("customer_id").n_unique().alias("customer_id_distintos"),
        pl.col("customer_unique_id").n_unique().alias("personas_distintas"),
    )
)
````

#### 2.2.2 — Montos admisibles

Construyan `items_silver` y `payments_silver`. Antes de conservar las tablas,
busquen precios, fletes y pagos negativos, `NaN` o infinitos. Si aparece uno,
levanten `ContractViolation`; no lo reemplacen por cero ni eliminen la fila.
Al terminar, muestren el mínimo de `price`, `freight_value` y `payment_value`.

````python
MONTOS = {
    "order_items": (items_raw, ("price", "freight_value")),
    "payments": (payments_raw, ("payment_value",)),
}

# Un NaN o un infinito no es "un monto raro": es un valor que rompe cualquier
# suma posterior. Un negativo tampoco se imputa a cero, porque eso inventaría
# evidencia. En ambos casos Silver se detiene.
auditoria_montos = pl.concat(
    [
        tabla.select(
            pl.lit(fuente).alias("fuente"),
            pl.lit(columna).alias("columna"),
            pl.col(columna).min().alias("minimo"),
            pl.col(columna).max().alias("maximo"),
            pl.col(columna).null_count().alias("nulos"),
            (pl.col(columna).is_not_null() & ~pl.col(columna).is_finite())
            .sum()
            .alias("no_finitos"),
            (pl.col(columna) < 0).sum().alias("negativos"),
        )
        for fuente, (tabla, columnas) in MONTOS.items()
        for columna in columnas
    ]
)
display(auditoria_montos)

if auditoria_montos["no_finitos"].sum() > 0:
    raise ContractViolation("Se recibieron montos no finitos.")
if auditoria_montos["negativos"].sum() > 0:
    raise ContractViolation("Se recibieron montos negativos.")

items_silver = items_raw
payments_silver = payments_raw

print(
    "Mínimos admitidos -> "
    f"price: {items_silver['price'].min()}; "
    f"freight_value: {items_silver['freight_value'].min()}; "
    f"payment_value: {payments_silver['payment_value'].min()}"
)
````

#### 2.2.3 — Relaciones que permiten publicar Silver

Auditen las tres relaciones con un `anti join`: `orders.customer_id` contra
`customers.customer_id`, `order_items.order_id` contra `orders.order_id` y
`payments.order_id` contra `orders.order_id`. Construyan una tabla con el
número de filas huérfanas de cada relación. Si alguno es mayor que cero,
levanten `ContractViolation` e indiquen la relación fallida.

````python
# El anti join devuelve las filas de la izquierda cuya llave no existe a la
# derecha. Cada una de estas tres relaciones debe dar cero para poder publicar.
RELACIONES = (
    ("orders.customer_id -> customers.customer_id", "customer_id", orders_silver, customers_silver),
    ("order_items.order_id -> orders.order_id", "order_id", items_silver, orders_silver),
    ("payments.order_id -> orders.order_id", "order_id", payments_silver, orders_silver),
)

auditoria_relaciones = pl.DataFrame(
    {
        "relacion": [etiqueta for etiqueta, _, _, _ in RELACIONES],
        "filas_huerfanas": [
            izquierda.join(
                derecha.select(llave).unique(), on=llave, how="anti"
            ).height
            for _, llave, izquierda, derecha in RELACIONES
        ],
    }
)
display(auditoria_relaciones)

fallidas = auditoria_relaciones.filter(pl.col("filas_huerfanas") > 0)
if fallidas.height:
    raise ContractViolation(
        "Clave foránea huérfana en: "
        f"{', '.join(fallidas['relacion'].to_list())}"
    )

print("Las tres relaciones Silver son válidas: se puede publicar la capa.")
````

#### 3.2.1 — Ventas diarias

Construyan `sales_daily`. Consideren solo órdenes `delivered`, únanlas con sus
ítems por `order_id` y creen `sale_date` desde la fecha de compra. El resultado
debe tener una fila por `sale_date`, `items_sold_value` como suma de `price` y
`delivered_orders` como número de `order_id` distintos. Los fletes no forman
parte de esta métrica.

````python
# Solo órdenes entregadas; sale_date desde la fecha de compra; los fletes no
# forman parte de la métrica. El join con ítems es 1:N, así que delivered_orders
# se cuenta con n_unique y no con len().
sales_daily = (
    orders_silver.filter(pl.col("order_status") == "delivered")
    .select(
        "order_id",
        pl.col("order_purchase_timestamp").dt.date().alias("sale_date"),
    )
    .join(items_silver.select("order_id", "price"), on="order_id", how="inner")
    .group_by("sale_date")
    .agg(
        pl.col("price").sum().alias("items_sold_value"),
        pl.col("order_id").n_unique().cast(pl.Int64).alias("delivered_orders"),
    )
    .sort("sale_date")
)

print(f"Días con ventas: {sales_daily.height}")
display(sales_daily.head())
display(
    sales_daily.select(
        pl.col("items_sold_value").sum().alias("valor_total_items"),
        pl.col("delivered_orders").sum().alias("ordenes_entregadas"),
    )
)
````

#### 3.2.2 — Exclusiones y variables RFM por persona

Conviertan primero `ordenes_sin_pago` en `rfm_exclusions`: una fila por
`order_id`, las columnas que ya conservaron y `reason` igual a
`delivered_order_without_payment`. Esa tabla no entra al RFM ni transforma un
pago faltante en cero.

Luego construyan `customer_rfm` solo desde órdenes entregadas con pago
registrado. Antes de unir pagos a clientes, agréguenlos por `order_id`; una
orden puede tener más de un pago y sumar después de un join al grano de cliente
duplicaría `Monetary`. Luego relacionen las órdenes con `customer_unique_id`,
agrupen por esa identidad y calculen:

- `last_purchase`: última fecha de compra elegible;
- `frequency`: número de órdenes elegibles distintas;
- `monetary`: suma de pagos de esas órdenes;
- `recency_days`: días calendario entre `last_purchase` y el día posterior a
  la última compra elegible del conjunto;
- `segment`: segmento calculado con `config/rfm_segments.yaml`.

No usen compras posteriores a la fecha de referencia ni incluyan una orden de
`rfm_exclusions`.

````python
from datetime import timedelta

import yaml

segments = yaml.safe_load(
    (RUTA_LAB / "config/rfm_segments.yaml").read_text()
)["segments"]

# --- Exclusiones: la auditoría de 3.1 más el motivo, sin recalcular nada. ---
rfm_exclusions = ordenes_sin_pago.with_columns(
    pl.lit("delivered_order_without_payment").alias("reason")
)
display(rfm_exclusions)

# --- Pagos al grano de orden ANTES de cualquier unión hacia cliente. ---
pagos_por_orden = payments_silver.group_by("order_id").agg(
    pl.col("payment_value").sum().alias("order_payment_value")
)

# El inner join con pagos deja fuera, por construcción, las órdenes de
# rfm_exclusions: son exactamente las entregadas sin pago.
ordenes_elegibles = (
    orders_silver.filter(pl.col("order_status") == "delivered")
    .join(pagos_por_orden, on="order_id", how="inner")
    .join(customers_silver, on="customer_id", how="inner")
)

# Fecha de referencia: el día posterior a la última compra elegible. Nada
# posterior a esta fecha entra al cálculo, para no filtrar información.
reference_date = ordenes_elegibles[
    "order_purchase_timestamp"
].dt.date().max() + timedelta(days=1)
print(f"Fecha de referencia: {reference_date}")


def expresion_segmento(reglas: dict) -> pl.Expr:
    """Traduce los umbrales congelados del YAML a una sola expresión."""
    comparaciones = {
        "max_recency_days": lambda v: pl.col("recency_days") <= v,
        "min_recency_days": lambda v: pl.col("recency_days") >= v,
        "min_frequency": lambda v: pl.col("frequency") >= v,
        "min_monetary": lambda v: pl.col("monetary") >= v,
    }
    expresion = pl.lit("Others")
    for nombre, umbrales in reversed(list(reglas.items())):
        condicion = pl.lit(True)
        for regla, valor in umbrales.items():
            condicion = condicion & comparaciones[regla](valor)
        expresion = (
            pl.when(condicion)
            .then(pl.lit(nombre.replace("_", " ").title()))
            .otherwise(expresion)
        )
    return expresion.alias("segment")


customer_rfm = (
    ordenes_elegibles.group_by("customer_unique_id")
    .agg(
        pl.col("order_purchase_timestamp").max().alias("last_purchase"),
        pl.col("order_id").n_unique().cast(pl.Int64).alias("frequency"),
        pl.col("order_payment_value").sum().alias("monetary"),
    )
    .with_columns(
        pl.lit(reference_date, dtype=pl.Date).alias("reference_date"),
        (
            pl.lit(reference_date, dtype=pl.Date)
            - pl.col("last_purchase").dt.date()
        )
        .dt.total_days()
        .cast(pl.Int64)
        .alias("recency_days"),
    )
    .with_columns(expresion_segmento(segments))
    .sort("customer_unique_id")
)

print(f"Personas en el RFM: {customer_rfm.height}")
display(customer_rfm.head())
display(
    customer_rfm["segment"].value_counts().sort("count", descending=True)
)
````
